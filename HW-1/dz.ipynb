```python
from IPython.display import HTML
import matplotlib.pyplot as plt
import matplotlib.animation as animation
import numpy as np
import glob
from collections import defaultdict 
import pickle

from draw import *

with open('LipsArrayHalfTime.np', 'rb') as f:
    LipsData = pickle.load(f)

print('Загруженные буквы:', list(LipsData.keys()))
```

```python
def lagrange_interpolate(t_old, y_old, t_new):
    n = len(t_old)
    y_new = np.zeros(len(t_new))
    for i in range(n):
        L = np.ones(len(t_new))
        for j in range(n):
            if i != j:
                L *= (t_new - t_old[j]) / (t_old[i] - t_old[j])
        y_new += y_old[i] * L
    return y_new
```

```python
Task1LispPoints = {}
for letter, frames in LipsData.items():
    n = frames.shape[0]
    t_old = np.arange(n)
    t_new = np.arange(0, n - 0.5, 0.5) 
    n_new = len(t_new) 
    frames_flat = frames.reshape(n, 80)
    new_frames_flat = np.zeros((n_new, 80))
    for coord_idx in range(80):
        y_old = frames_flat[:, coord_idx]
        new_frames_flat[:, coord_idx] = lagrange_interpolate(t_old, y_old, t_new)
    Task1LispPoints[letter] = new_frames_flat.reshape(n_new, 40, 2)
```
